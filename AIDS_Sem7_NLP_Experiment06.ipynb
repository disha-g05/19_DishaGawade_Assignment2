Import Libraries

```python
print("Disha Gawade 231A018")
import nltk
from nltk.tokenize import word_tokenize
from nltk.tag import UnigramTagger, BigramTagger, pos_tag
from nltk.corpus import treebank
from nltk import pos_tag_sents
```

```python
print("Disha Gawade 231A018")
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('punkt_tab')
nltk.download('treebank')
nltk.download('tagsets_json')
```

```python
# Tagging a sample text
print("Disha Gawade 231A018")
sample_text = "He is reading a book. Book a flight ticket for me."

# Tokenize the text
tokens = word_tokenize(sample_text)
```

```python
print("Disha Gawade 231A018")
# Rule based POS tagging
rule_based_tags = pos_tag(tokens)
```

```python
print("Disha Gawade 231A018")
print(rule_based_tags)
```

```python
print("Disha Gawade 231A018")
nltk.help.upenn_tagset('VBG')
```

```python
print("Disha Gawade 231A018")
# Statistical tagging using treebank corpus
train_sents = treebank.tagged_sents()
# Unigram tagger
unigram_tagger = UnigramTagger(train_sents)
unigram_tags = unigram_tagger.tag(tokens)

# Bigram tagger
bigram_tagger = BigramTagger(train_sents, backoff=unigram_tagger)
bigram_tags = bigram_tagger.tag(tokens)
```

```python
print("Disha Gawade 231A018")
bigram_tags
```

```python
print("Disha Gawade 231A018")
unigram_tags
```

```python
print("Disha Gawade 231A018")
print("Original Sentence: ", sample_text)
print("\nRule_Based POS Tagging (statistical):")
print(rule_based_tags)
print("\nUnigram POS Tagging")
print(unigram_tags)
print("\nBigram POS Tagging")
print(bigram_tags)
```

```python
print("Disha Gawade 231A018")
# Evaluate the performance of each tagger using Treebank
test_sents = treebank.tagged_sents()[1000:]
unigram_acc = unigram_tagger.evaluate(test_sents)
bigram_acc = bigram_tagger.evaluate(test_sents)

print(f"\nUnigram tagger accuracy: {unigram_acc * 100: .2f}%")
print(f"\nBigram tagger accuracy: {bigram_acc * 100: .2f}%")
```
