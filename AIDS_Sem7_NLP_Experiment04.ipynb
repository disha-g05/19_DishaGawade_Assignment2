```python
print("Disha Gawade, 231A018")
text = "The cats are running and jumping over the fences."
```

```python
print("Disha Gawade, 231A018")
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
import string
```

```python
print("Disha Gawade, 231A018")
# Download resouces
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('punkt_tab')
```

```python
print("Disha Gawade, 231A018")
import string
print(string.punctuation)
```

```python
# input text
# Step 1: preprocessing
print("Disha Gawade, 231A018")
text_lower = text.lower()
print(text_lower)
```

```python
# Step 2: Tokenization
print("Disha Gawade, 231A018")
tokens = word_tokenize(text_lower)

print(tokens)
```

```python
print("Disha Gawade, 231A018")
text_clean=[]
for word in tokens:
  if word not in string.punctuation:
    text_clean.append(word)
print(text_clean)
```

```python
# Step 3: Identify roots (morpheme approximation using lemmatizer)
print("Disha Gawade, 231A018")
lemmatizer = WordNetLemmatizer()
lemmas = [lemmatizer.lemmatize(tok, pos='v') for tok in text_clean] #lemmatize as verbs

lemmas
```

```python
print("Disha Gawade, 231A018")
[(token, lemma) for token, lemma in zip(tokens,lemmas)]
```

```python
# Step 4: Analyze tokens manually for derivational vs inflectional
print("Disha Gawade, 231A018")
morphological_info = {}
for token, lemma in zip(tokens, lemmas):
  if token != lemma:
    # This indicates some morphological change
    if token.endswith("ing"):
      analysis = "Inflectional: Present participle/continuous form of '{}'".format(lemma)
    elif token.endswith("er") or token.endswith("ation"):
      analysis = "Derivational: '{}' + -er/-ation -> modified form".format(lemma)
    else:
      analysis = "Inflectional or irregular form of '{}'".format(lemma)
  else:
    analysis = "Base/root form"
  morphological_info[token]=analysis
```

```python
print("Disha Gawade, 231A018")
morphological_info
```
